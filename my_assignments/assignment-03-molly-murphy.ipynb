### Task 12

In two or three sentences: what is the difference between cloning a repository and forking it? Give one situation where you need a fork.

*Cloning a repository downloads it from GitHub to your computer, while forking one makes a copy of it on your GitHub account. You would need a fork when you want to build something new from a repository by being able to push changes to your own copy without having write access to the upstream repository.
