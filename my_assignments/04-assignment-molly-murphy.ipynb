Run this setup:

```bash
mkdir secret-practice
cd secret-practice
git init -b main
echo "api_key=12345" > secrets.txt
git add secrets.txt
git commit -m "Add config"
```

**ls**<br><br>
*Output:*<br>
![secret](<attachment:Screenshot 2026-09-29 at 3.18.34 PM.png>)<br><br>
