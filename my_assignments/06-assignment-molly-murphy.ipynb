### Task 01

In your own words, explain why the vector arithmetic king − man + woman ≈ queen works. Your answer must use the terms "embedding" and "direction" and be no longer than four sentences.

Then propose one analogy of your own in the same form (a − b + c ≈ d), different from the three on the slides, and state in one sentence what the direction a − b represents.

**The vector arithmetic king − man + woman ≈ queen works because the words are tokens that become embeddings, which are numbers representing coordinates in space. Tokens used in similar ways are closer together in the embedding space, with directions carrying meaning, which allows for relationships between meanings to become vector operations. Here, the direction roughly means gender, so when you take "king", subtract "man" from it, and add "woman", "queen" is the nearest word or token in the embedding space.**<br><br>

**Another analogy could be tennis − court + ice rink ≈ hockey. Here, the direction tennis - court represents a sport's playing surface.**
