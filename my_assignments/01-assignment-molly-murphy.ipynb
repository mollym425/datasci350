### Question 11

Describe the Von Neumann bottleneck in two sentences, and name one hardware feature that reduces its impact.

**The Von Neumann bottleneck occurs because the wires connecting the CPU to memory can only move either instructions or data at once since they carry both. This results in the CPU waiting for memory, causing it to be inactive while it waits. Caches are a hardware feature that reduces its impact.**
